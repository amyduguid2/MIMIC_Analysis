```python
# Wrap your dataframe into a SampleDataset
dataset = SampleEHRDataset(dataset_name="mortality_demo")

# Add samples (each row is a visit/admission)
for _, row in df.iterrows():
    dataset.add_sample(
        patient_id=str(row["subject_id"]),
        visit_id=str(row["hadm_id"]),
        codes={"diagnoses": row["ccs_seq"]},   # must be a list[str]
        label=row["hospital_expire_flag"],
        attributes={
            "gender": row["gender"],
            "race": row["race"],
            "insurance": row["insurance"],
            "admission_type": row["admission_type"],
            "age": row["age_at_admission"],
            "los": row["admission_duration"],  # for LOS regression
            # … add engineered features here
        }
    )
```
